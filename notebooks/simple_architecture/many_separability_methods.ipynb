```python
timings = {}

@contextmanager
def timer(name: str):
    start = time.perf_counter()
    try:
        yield
    finally:
        elapsed = time.perf_counter() - start
        timings[name] = elapsed
        print(f"[TIMER] {name}: {elapsed:.4f} s")
```

### 1. Подготовка данных:

```python
dataset_name = "cifar10"
dataset_config = get_dataset_config(dataset_name)
all_class_ids = list(range(dataset_config.num_classes))

transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(dataset_config.mean, dataset_config.std)
])


generator = torch.Generator().manual_seed(seed)
output_dir = ensure_dir(OUTPUT_DIR / dataset_name / str(target)) 


### То, на чем обучаем resnet, строим pca
train_dataset = build_excluded_dataset(
    name=dataset_name,
    root=str(DATA_DIR),
    exclude_class=[target],
    train=True, 
    transform=transform
)

train_size = int(0.8 * len(train_dataset))
val_size = int(0.15 * len(train_dataset))

# выборка для построение PCA пространства, для тренеровки модели разделения
test_size = (len(train_dataset) - train_size - val_size) // 2 
# выбока, которая отрпаится в финальный тест нового пайплайна
new_pipeline_test_size = len(train_dataset) - train_size - val_size - test_size


train, val, test, new_pipeline_test = random_split(
    train_dataset, 
    [train_size, val_size, test_size, new_pipeline_test_size], 
    generator=generator
)

train_loader = DataLoader(train, batch_size=64, shuffle=True, num_workers=2, generator=generator)
val_loader   = DataLoader(val, batch_size=64, shuffle=False, num_workers=2, generator=generator)
test_loader = DataLoader(test, batch_size=64, shuffle=False, num_workers=2, generator=generator)

### То, на чем делаем холостой (все классы будует неверные) предикт ради feature layer, на чем обучаем и проводим тесты lda
target_dataset = build_excluded_dataset(
    name=dataset_name,
    root=str(DATA_DIR),
    exclude_class=[class_id for class_id in all_class_ids if class_id != target],
    train=True, 
    transform=transform
)

target_train_size = int(0.85 * len(target_dataset))
target_test_size = (len(target_dataset) - target_train_size)
other = len(target_dataset) - target_train_size - target_test_size

train_target, test_target, other = random_split(
    target_dataset, 
    [target_train_size, target_test_size, other],
    generator=generator
)

train_target_loader = DataLoader(train_target, batch_size=64, shuffle=True, num_workers=2, generator=generator)
test_target_loader = DataLoader(test_target, batch_size=64, shuffle=True, num_workers=2, generator=generator)

### То, на чем пробуем новый пайплайн
new_pipeline_loader = DataLoader(new_pipeline_test+test_target, batch_size=64, shuffle=True, num_workers=2, generator=generator)
```

```python
all_class_ids = list(range(dataset_config.num_classes))
features, preds, gt_preds = extract_features(resnet.model, new_pipeline_loader, device)
_baseline_features, baseline_preds, baseline_gt_preds = extract_features(baseline_resnet.model, new_pipeline_loader, device)
baseline_f1_macro = f1_score(baseline_gt_preds, baseline_preds, average="macro", labels=all_class_ids, zero_division=0)
baseline_f1_list = f1_score(baseline_gt_preds, baseline_preds, average=None, labels=all_class_ids, zero_division=0)

print(f"Baseline on new pipeline ({target} / {target_name}) f1-macro: {baseline_f1_macro}")
print(f"Baseline per-class f1: {baseline_f1_list}")

mlflow.log_metric("baseline_new_pipeline_f1_macro", baseline_f1_macro)
for cls_id, cls_name, f1 in zip(all_class_ids, cifar10_class_names, baseline_f1_list):
    mlflow.log_metric(f"baseline_new_pipeline_f1_class_{cls_id}_{cls_name}", f1)

def benchmark_step(step_fn, repeats=100, warmup=10):
    for _ in range(warmup):
        step_fn()

    timings = []
    result = None
    for _ in range(repeats):
        start = time.perf_counter()
        result = step_fn()
        timings.append(time.perf_counter() - start)

    timings = np.asarray(timings, dtype=float)
    return result, timings.mean(), timings.std(ddof=1) if repeats > 1 else 0.0, np.median(timings)

def new_pipeline(features: np.array, preds: np.array, sep_model, proj, repeats=100, warmup=10):
    features_p, transform_time_sec, transform_std_sec, transform_median_sec = benchmark_step(
        lambda: proj.transform(features),
        repeats=repeats,
        warmup=warmup,
    )

    # se_predict =  np.where(lda.predict_proba(features_p)[:, 1] > thresh, 1, 0)
    sep_predict, sep_predict_time_sec, sep_predict_std_sec, sep_predict_median_sec = benchmark_step(
        lambda: sep_model.predict(features_p),
        repeats=repeats,
        warmup=warmup,
    )

    def build_prediction():
        return np.array([target if sep_predict[i] else preds[i] for i in range(sep_predict.shape[0])])

    predict, postprocess_time_sec, postprocess_std_sec, postprocess_median_sec = benchmark_step(
        build_prediction,
        repeats=repeats,
        warmup=warmup,
    )

    return (
        predict,
        transform_time_sec, transform_std_sec, transform_median_sec,
        sep_predict_time_sec, sep_predict_std_sec, sep_predict_median_sec,
        postprocess_time_sec, postprocess_std_sec, postprocess_median_sec,
    )

def print_metric(preds, gt_preds, baseline_f1_list=None, baseline_f1_macro=None):
    # https://scikit-learn.org/stable/modules/generated/sklearn.metrics.f1_score.html
    new_pipeline_loader_f1_score_macro = f1_score(gt_preds, preds, average="macro", labels=all_class_ids, zero_division=0)
    new_pipeline_loader_f1_score_micro = f1_score(gt_preds, preds, average="micro", labels=all_class_ids, zero_division=0)
    new_pipeline_loader_f1_score_list = f1_score(gt_preds, preds, average=None, labels=all_class_ids, zero_division=0)
    new_pipeline_loader_acc = accuracy_score(gt_preds, preds)
    power_list = None
    power_macro = None

    print("f1-мера (глобальные TP и FP): {} \nf1-score (усреднее): {}" \
    "\nf1-score (по каждому классу): {} \nacc: {}".format(
        new_pipeline_loader_f1_score_macro,
        new_pipeline_loader_f1_score_micro,
        new_pipeline_loader_f1_score_list,
        new_pipeline_loader_acc
        )
    )

    if baseline_f1_macro is not None and baseline_f1_macro != 0:
        power_macro = new_pipeline_loader_f1_score_macro / baseline_f1_macro
        print(f"Power macro ({target} / {target_name}): {power_macro}")

    if baseline_f1_list is not None:
        power_list = np.divide(
            new_pipeline_loader_f1_score_list,
            baseline_f1_list,
            out=np.full_like(new_pipeline_loader_f1_score_list, np.nan, dtype=float),
            where=baseline_f1_list != 0,
        )
        print(f"Power per class ({target} / {target_name}): {power_list}")

    return new_pipeline_loader_f1_score_macro, new_pipeline_loader_f1_score_list, power_macro, power_list

def evaluate_and_log_method(method_name, sep_model, proj):
    method_start = time.perf_counter()
    (
        preds_pca,
        transform_time_sec, transform_std_sec, transform_median_sec,
        sep_predict_time_sec, sep_predict_std_sec, sep_predict_median_sec,
        postprocess_time_sec, postprocess_std_sec, postprocess_median_sec,
    ) = new_pipeline(features, preds, sep_model, proj)
    new_pipeline_f1_macro, new_pipeline_f1_list, power_macro, power_list = print_metric(
        preds_pca,
        gt_preds,
        baseline_f1_list=baseline_f1_list,
        baseline_f1_macro=baseline_f1_macro,
    )

    new_pipeline_conf_m = plot_confusion_matrix(
        preds_pca,
        gt_preds,
        title=f"Confusion Matrix with {method_name} sep for {target} ({target_name})",
        path=str(output_dir / method_name / "new_pipeline_confusion_matrix.png")
    )

    mlflow.log_artifact(new_pipeline_conf_m, method_name)
    mlflow.log_metric(f"new_pipeline_{method_name}_f1_macro", new_pipeline_f1_macro)
    mlflow.log_metric(f"new_pipeline_{method_name}_Power_macro", power_macro)
    mlflow.log_metric(f"new_pipeline_{method_name}_transform_time_ms", transform_time_sec * 1000.0)
    mlflow.log_metric(f"new_pipeline_{method_name}_transform_std_ms", transform_std_sec * 1000.0)
    mlflow.log_metric(f"new_pipeline_{method_name}_transform_median_ms", transform_median_sec * 1000.0)
    mlflow.log_metric(f"new_pipeline_{method_name}_sep_predict_time_ms", sep_predict_time_sec * 1000.0)
    mlflow.log_metric(f"new_pipeline_{method_name}_sep_predict_std_ms", sep_predict_std_sec * 1000.0)
    mlflow.log_metric(f"new_pipeline_{method_name}_sep_predict_median_ms", sep_predict_median_sec * 1000.0)
    mlflow.log_metric(f"new_pipeline_{method_name}_postprocess_time_ms", postprocess_time_sec * 1000.0)
    mlflow.log_metric(f"new_pipeline_{method_name}_postprocess_std_ms", postprocess_std_sec * 1000.0)
    mlflow.log_metric(f"new_pipeline_{method_name}_postprocess_median_ms", postprocess_median_sec * 1000.0)
    method_elapsed_sec = time.perf_counter() - method_start
    print(f"[{method_name}] elapsed: {method_elapsed_sec:.4f} s")
    mlflow.log_metric(f"new_pipeline_{method_name}_time_sec", method_elapsed_sec)
    for cls_id, cls_name, f1, power in zip(all_class_ids, cifar10_class_names, new_pipeline_f1_list, power_list):
        mlflow.log_metric(f"new_pipeline_{method_name}_f1_class_{cls_id}_{cls_name}", f1)
        mlflow.log_metric(f"new_pipeline_{method_name}_Power_class_{cls_id}_{cls_name}", power)

    return preds_pca, new_pipeline_f1_macro, new_pipeline_f1_list, power_macro, power_list
```
